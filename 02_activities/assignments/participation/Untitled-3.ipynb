```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
fig, ax = plt.subplots(figsize = (5,3))
ax.plot (x,y1)
ax.plot (x,y2)
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label="Person 2")
ax.legend(loc='lower right')
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label="Person 2")
ax.legend(loc='upper center')
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label="Person 2")
ax.legend(loc='best')
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label="Person 2")
ax.legend(loc='lower right',
          frameon = True, #add frame around the legend
          fontsize = 12, #change fontsize
          ncol=2, #specify number of columns
          shadow = True)
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.plot(x,y1, label="Person 1")
ax.plot(x,y2, label="Person 2")
ax.legend(loc='upper left',
          bbox_to_anchor = (1,1))
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label="Person 2")
ax.legend(loc='lower right')
ax.text(10,95,"This value is important")
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label="Person 2")
ax.legend(loc='lower right', prop={"family":"serif", "size":20})
ax.text(10,95,"This value is important",
        ha='center', # alignment of text
        color = 'red', # modify font color
        size = 20 ) # modify font size
fig.show()
```

```python
fig, ax=plt.subplots()
ax.axis ([0,10,0,10])
```

```python
fig, ax=plt.subplots()
ax.axis ([0,10,0,10])
ax.text(1,5,". Data:(1,5)", transform=ax.transData) # data coordinate
ax.text(0.5,0.1,". Data:(0.5,0.1)", transform=ax.transAxes) # 50% of x-axis and 10% of y-axis
ax.text(0.2,0.2,". Data:(0.2,0.2)", transform=fig.transFigure) # 20% of the whole figure
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label="Person 2")
ax.legend(loc='lower right')
ax.annotate ('this is important!', 
             xy=(10,95),
             xytext = (20,94), 
             arrowprops=dict(facecolor='black'))
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label="Person 2")
ax.legend(loc='lower right')
ax.annotate ('this is important!', 
             xy=(10,95),
             xytext = (20,94), 
             color = 'indigo',
             arrowprops=dict(arrowstyle="fancy",
                             color='hotpink'))
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label="Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator()) # y-axis gone
ax.xaxis.set_major_formatter(plt.NullFormatter()) # x-axis number gone
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label="Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label="Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
fig.show()
```

```python
fig, ax= plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label="Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right')
fig.show()
```
